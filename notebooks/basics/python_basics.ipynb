```python
class dog:
    def __init__(self, name, age):
        self.name=name
        self.age=age
    
    def bark(self,activity):
        return f"{self.name} {self.age} is {activity}"
    
my_dog = dog("Dog", 3)
print(my_dog.bark("playing"))
```
